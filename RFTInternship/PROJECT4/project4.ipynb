```python
LOGS = [
    "ERROR DISK FULL",
    "INFO STARTED",
    "ERROR FILE MISSING",
    "WARNING MEMORY LOW"
]
```

```python
def analyze_logs(logs):
    
    counts = {
        "ERROR": 0,
        "INFO": 0,
        "WARNING": 0        
    }

    for log in logs:
        if "ERROR" in log.upper():
            counts["ERROR"] += 1
        elif "INFO" in log.upper():
            counts["INFO"] += 1
        elif "WARNING" in log.upper():  
            counts["WARNING"] += 1

    print("Log Analysis:")
    for log_type, count in counts.items():
        print(f"{log_type}: {count}")
    
    most_freq = max(counts, key=counts.get)
    print(f"Most frequent log type: {most_freq}")

analyze_logs(LOGS)
```
