```python
import pandas as pd
from io import StringIO

data = """
PRODUCT,QUANTITY,PRICE
A,2,100
B,1,200
C,5,50"""
```

```python
# Read CSV data
df = pd.read_csv(StringIO(data))

# Add new column: TOTAL = QUANTITY * PRICE
df["TOTAL"] = df["QUANTITY"] * df["PRICE"]

print("Dataset:\n")
print(df)

# Calculate total sales per product
sales_per_product = df.groupby("PRODUCT")["TOTAL"].sum()

print("\nTotal Sales Per Product:\n")
print(sales_per_product)

# Find total revenue
total_revenue = df["TOTAL"].sum()
print("\nTotal Revenue:", total_revenue)

# Find top-selling product
top_selling = sales_per_product.idxmax()
print("Top-Selling Product:", top_selling)

# Sort by revenue
sorted_sales = sales_per_product.sort_values(ascending=False)

print("\nSorted by Revenue:\n")
print(sorted_sales)
```
