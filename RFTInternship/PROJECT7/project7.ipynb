```python
import numpy as np
data = {
    "NAME" : ['AMIT', 'RIYA', 'JOHN'],
    "MATHS" : [80, 90, 60],
    "SCIENCE" : [70, 88, 65],
    "ENGLISH" : [85, 92, 70]
}
```

```python
import pandas as pd

# Create a DataFrame from the data dictionary
df = pd.DataFrame(data, index = ['STUDENT1', 'STUDENT2', 'STUDENT3'])
df["TOTAL"] = df["MATHS"] + df["SCIENCE"] + df["ENGLISH"]

total = len(data["NAME"])
#finding the average of each student 
df["AVERAGE"] = df["TOTAL"] / total

#finding the topper student
topper = df["TOTAL"].idxmax()


#counting the students with average above 75
above = len(df[df["AVERAGE"] > 75])

    

#adding grade column based on average marks
df["GRADE"] = np.where(df["AVERAGE"] >= 90, 'A',
                       np.where(df["AVERAGE"] >= 80, 'B',
                                np.where(df["AVERAGE"] >= 70, 'C', 'D')))

#subject-wise average
sub_avg = df[["MATHS", "SCIENCE", "ENGLISH"]].mean()


print("STUDENTS MARKS:")
print(df, '\n')
print("Topper Student:", topper, '\n')
print("Students with average above 75:", above, '\n')
print("Subject-wise Average:")
print(sub_avg)
```
