```python
dataset = {
    "NAME" : ["A", "B", "C", "D"],
    "DEPT" : ["IT", "HR", "IT", "HR"],
    "SALARY" : [50000, 40000, 60000, 45000]
}
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(dataset, index = ["employee1", "employee2", "employee3", "employee4"])

#finding average salary per department 
avg_salary = df.groupby("DEPT")["SALARY"].mean()

#finding highest paid employee per department
high_paid_employee = df.groupby("DEPT")["SALARY"].idxmax()

#count employee in each department
employee_count = df.groupby("DEPT")["NAME"].count()

#sort deptartment by average salary in descending order
avg_dept = avg_salary.sort_values(ascending = False)

print(df)
print("\n" , avg_salary)
print("\n" , "Highest paid employee per department:","\n" , df["NAME"].loc[high_paid_employee])
print("\n" , "Employee count per department:","\n" , employee_count)
print("\n" , "Average salary per department (sorted):","\n" , avg_dept)
```
