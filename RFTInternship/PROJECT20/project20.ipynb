```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
# 1. Import Data (Built-in restaurant sales/tips data)
df = sns.load_dataset('tips')

# 2. Data Cleaning (Check for missing values and filter out negative anomalies if any)
df = df.dropna()
df = df[df['total_bill'] > 0]  # Ensure bills are valid numbers

# 3. Analysis (Calculate simple group averages)
# Let's see who tips better on average: Smokers or Non-Smokers?
print("--- Average Tip Amount by Smoker Status ---")
print(df.groupby('smoker')['tip'].mean())

print("\n--- Average Total Bill by Day of the Week ---")
print(df.groupby('day')['total_bill'].mean())

# 4. Visualization (Simple charts side-by-side)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Chart 1: Bar plot of total bill per day
sns.barplot(data=df, x='day', y='total_bill', ax=ax1, palette='viridis', errorbar=None)
ax1.set_title('Average Total Bill per Day')
ax1.set_xlabel('Day of the Week')
ax1.set_ylabel('Total Bill ($)')

# Chart 2: Scatter plot showing Relationship between Bill vs Tip
sns.scatterplot(data=df, x='total_bill', y='tip', hue='time', ax=ax2)
ax2.set_title('Total Bill vs Tip Amount')
ax2.set_xlabel('Total Bill ($)')
ax2.set_ylabel('Tip Given ($)')

plt.tight_layout()
plt.show()
```
