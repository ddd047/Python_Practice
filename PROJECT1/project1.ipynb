```python
data = ["10", None, "20", "10", " ", "30", None, "40"]
clean = []
junk = []
print(data)
```

```python
# removing invalid values (None, emplty string)
for x in data[:] :
    if x is None or not x.strip():
        junk.append(x)
    else :
        clean.append(x)

print(clean)
```

```python
# counting how many values were removed 

count = 0
for _ in junk:
    count += 1

print(count)
```

```python
# sorting the final list 

n = len(clean)

for i in range(n) :
    for j in range(0, n-i-1) :
        if clean[j] > clean[j+1] :
            clean[j], clean[j+1] = clean[j+1], clean[j]

print(clean)
```
