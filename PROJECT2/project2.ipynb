```python
Marks = [78, 85, 90, 67, 85, 92, 78]
```

```python
# Calculate the average mark
average_mark = sum(Marks) / len(Marks)
print(f"The average mark is: {average_mark:.2f}")
```

```python
#calculate the highest mark and lowest mark
highest_mark = max(Marks)
lowest_mark = min(Marks)
print(f"The highest mark is: {highest_mark}")
print(f"The lowest mark is: {lowest_mark}")
```

```python
#count how many students scored above average
above_average_count = sum(1 for mark in Marks if mark > average_mark)
print(f"Number of students who scored above average: {above_average_count}")
```

```python
#Create grade distribution
for x in Marks:
    if x >= 85:
        print(f"{x} - A")
    elif x >= 75:
        print(f"{x} - B")
    elif x >= 65:
        print(f"{x} - C")
    elif x >= 55:
        print(f"{x} - D")
    else:
        print(f"{x} - F")
```
